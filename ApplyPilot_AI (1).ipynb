```python
import pandas as pd
from urllib.parse import urlsplit, urlunsplit

def clean_url(url):
    if not url:
        return ""
    parts = urlsplit(url)
    return urlunsplit((
        parts.scheme,
        parts.netloc.lower(),
        parts.path.rstrip("/"),
        "",
        ""
    ))

jobs = []
seen_urls = set()

for item in results:
    title = item.get("title", "").strip()
    url = item.get("url", "").strip()

    if not title or not url:
        continue

    normalized_url = clean_url(url)

    if normalized_url in seen_urls:
        continue

    seen_urls.add(normalized_url)

    jobs.append({
        "Title": title,
        "Company / Website": item.get("site_name", ""),
        "URL": url,
        "Description": item.get("snippet", "")
    })

jobs_df = pd.DataFrame(jobs)

print("Unique listings:", len(jobs_df))
display(jobs_df)
```

```python
from urllib.parse import urlparse

def looks_like_job_page(row):
    url = row["URL"].lower()
    title = row["Title"].lower()
    domain = urlparse(url).netloc

    # Exclude broad job-search and category pages
    broad_page_terms = [
        "/jobs/", "/internships/", "/job-search",
        "/q-", "/job/india-", "/search?"
    ]

    if any(term in url for term in broad_page_terms):
        return False

    # Prioritize common individual-job URL patterns
    individual_patterns = [
        "jobs.lever.co/",
        "boards.greenhouse.io/",
        "job-boards.greenhouse.io/",
        "jobs.ashbyhq.com/"
    ]

    if any(pattern in url for pattern in individual_patterns):
        return True

    # Keep other URLs for inspection if their titles
    # look like specific roles rather than collections.
    collection_terms = [
        "jobs in india", "internships in india",
        "job vacancies", "open jobs", "intern jobs"
    ]

    return not any(term in title for term in collection_terms)

jobs_df["Likely individual opening"] = jobs_df.apply(
    looks_like_job_page, axis=1
)

candidates_df = jobs_df[
    jobs_df["Likely individual opening"]
].copy()

print("Candidate URLs:", len(candidates_df))
display(candidates_df[["Title", "Website", "URL"]])
```

```python
import re
import pandas as pd

def extract_field(text, patterns):
    for pattern in patterns:
        match = re.search(pattern, text, re.I | re.M)
        if match:
            return match.group(1).strip()
    return "Not specified"

structured_jobs = []

for page in fetch_result.results:
    text = page.text or ""

    title = (
        re.search(r"^#\s+(.+)", text, re.M)
        or re.search(r"^TITLE:\s*(.+)", text, re.M)
    )

    title = title.group(1).strip() if title else page.title

    company = extract_field(text, [
        r"Company:\s*(.+)",
        r"About\s+([A-Z][A-Za-z0-9& .-]+)"
    ])

    location = extract_field(text, [
        r"Location:\s*(.+)",
        r"Locations?:\s*(.+)"
    ])

    structured_jobs.append({
        "Job Title": title,
        "Company": company,
        "Location": location,
        "Job URL": page.url,
        "Page Text Preview": re.sub(r"\s+", " ", text)[:350]
    })

structured_df = pd.DataFrame(structured_jobs)

print("Pages processed:", len(structured_df))
display(structured_df)
```

```python
import pandas as pd
import re
from urllib.parse import urlparse

def get_company_from_url(url):
    host = urlparse(url).netloc.lower()

    if host == "jobs.lever.co":
        parts = urlparse(url).path.strip("/").split("/")
        if parts:
            return parts[0].replace("-", " ").title()

    if host in ["boards.greenhouse.io", "job-boards.greenhouse.io"]:
        parts = urlparse(url).path.strip("/").split("/")
        if parts:
            return parts[0].replace("-", " ").title()

    return host.removeprefix("www.")

clean_jobs = []

for page in fetch_result.results:
    text = page.text or ""

    # Use the first Markdown heading as the job title
    match = re.search(r"^#\s+(.+)", text, re.MULTILINE)
    title = match.group(1).strip() if match else page.title

    clean_jobs.append({
        "Job Title": title,
        "Company": get_company_from_url(page.url),
        "Location": "Needs verification",
        "Job URL": page.url,
        "Description": re.sub(r"\s+", " ", text)[:500]
    })

clean_jobs_df = pd.DataFrame(clean_jobs)

display(clean_jobs_df)
```

```python
import re
import pandas as pd

def extract_location(text):
    patterns = [
        r"(?im)^Location\s*:\s*(.+)$",
        r"(?im)^Locations?\s*:\s*(.+)$",
        r"(?im)^Location\s*\n+\s*(.+)$",
        r"(?im)^Office location\s*:\s*(.+)$"
    ]

    for pattern in patterns:
        match = re.search(pattern, text)
        if match:
            value = match.group(1).strip()
            if len(value) < 100:
                return value

    return "Not found on page"

def extract_job_type(text):
    if re.search(r"\bintern(ship)?\b", text, re.I):
        return "Internship"
    if re.search(r"\bfull[- ]time\b", text, re.I):
        return "Full-time"
    if re.search(r"\bpart[- ]time\b", text, re.I):
        return "Part-time"
    return "Not specified"

improved_jobs = []

for page in fetch_result.results:
    text = page.text or ""
    row = clean_jobs_df.loc[
        clean_jobs_df["Job URL"] == page.url
    ]

    if row.empty:
        continue

    item = row.iloc[0].to_dict()
    item["Location"] = extract_location(text)
    item["Job Type"] = extract_job_type(text)
    item["Page Title"] = page.title
    improved_jobs.append(item)

improved_jobs_df = pd.DataFrame(improved_jobs)

display(improved_jobs_df[
    ["Job Title", "Company", "Location",
     "Job Type", "Job URL"]
])
```

```python
import re
import pandas as pd

# User preferences
preferred_roles = ["AI", "Machine Learning", "Data Science"]
preferred_locations = ["India", "Remote"]
preferred_skills = ["Python", "SQL", "Machine Learning"]
preferred_level = "Internship"

def calculate_match(job):
    title = str(job["Job Title"]).lower()
    description = str(job.get("Description", "")).lower()
    text = title + " " + description

    score = 0
    reasons = []

    # Role match: 40 points
    matched_roles = [
        role for role in preferred_roles
        if role.lower() in text
    ]
    if matched_roles:
        score += min(40, 20 * len(matched_roles))
        reasons.append("Roles: " + ", ".join(matched_roles))

    # Skill match: 40 points
    matched_skills = [
        skill for skill in preferred_skills
        if skill.lower() in text
    ]
    if matched_skills:
        score += min(40, 40 * len(matched_skills) / len(preferred_skills))
        reasons.append("Skills: " + ", ".join(matched_skills))

    # Level match: 20 points
    if preferred_level.lower() in text:
        score += 20
        reasons.append("Internship-level wording found")

    return pd.Series({
        "Match Score": round(score),
        "Match Reasons": "; ".join(reasons) if reasons else "Weak keyword match"
    })

match_results = improved_jobs_df.apply(calculate_match, axis=1)
ranked_jobs_df = pd.concat(
    [improved_jobs_df, match_results], axis=1
).sort_values("Match Score", ascending=False)

display(ranked_jobs_df[
    ["Job Title", "Company", "Match Score",
     "Match Reasons", "Job URL"]
])
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

role_input = widgets.Text(
    value="AI / Machine Learning Intern",
    description="Role:",
    placeholder="e.g. Data Scientist"
)

location_input = widgets.Text(
    value="India",
    description="Location:",
    placeholder="e.g. Remote, Bengaluru"
)

skills_input = widgets.Text(
    value="Python, SQL, Machine Learning",
    description="Skills:",
    placeholder="Enter comma-separated skills"
)

level_input = widgets.Dropdown(
    options=["Internship", "Entry-level", "Any"],
    value="Internship",
    description="Level:"
)

search_button = widgets.Button(
    description="Find Matching Jobs",
    button_style="success",
    icon="search"
)

output = widgets.Output()

def find_matching_jobs(button):
    with output:
        clear_output(wait=True)

        roles = [
            word.strip().lower()
            for word in role_input.value.split("/")
            if word.strip()
        ]

        locations = [
            word.strip().lower()
            for word in location_input.value.split(",")
            if word.strip()
        ]

        skills = [
            word.strip().lower()
            for word in skills_input.value.split(",")
            if word.strip()
        ]

        def score_job(job):
            title = str(job.get("Job Title", "")).lower()
            description = str(job.get("Description", "")).lower()
            location = str(job.get("Location", "")).lower()
            text = title + " " + description

            role_score = (
                sum(1 for role in roles if role in title)
                / max(len(roles), 1)
            )

            skill_score = (
                sum(1 for skill in skills if skill in text)
                / max(len(skills), 1)
            )

            location_score = (
                1.0 if any(loc in location for loc in locations) else 0.0
            )

            level_score = (
                1.0 if level_input.value == "Any"
                or level_input.value.lower() in text
                else 0.0
            )

            score = (
                0.40 * role_score
                + 0.35 * skill_score
                + 0.15 * location_score
                + 0.10 * level_score
            ) * 100

            return round(score)

        ranked = improved_jobs_df.copy()
        ranked["Match Score"] = ranked.apply(score_job, axis=1)
        ranked = ranked.sort_values("Match Score", ascending=False)

        display(ranked[
            ["Job Title", "Company", "Location",
             "Match Score", "Job URL"]
        ])

search_button.on_click(find_matching_jobs)

display(widgets.VBox([
    role_input,
    location_input,
    skills_input,
    level_input,
    search_button,
    output
]))
```

```python
import re
from urllib.parse import urlparse

def is_job_listing(page):
    url = page.url.lower()
    text = (page.text or "").lower()
    title = (page.title or "").lower()

    # Common individual job-page patterns
    individual_patterns = [
        "jobs.lever.co/",
        "boards.greenhouse.io/",
        "job-boards.greenhouse.io/",
        "jobs.ashbyhq.com/"
    ]

    if any(pattern in url for pattern in individual_patterns):
        return True

    # Look for evidence of an individual role
    role_terms = [
        "job description", "responsibilities",
        "qualifications", "requirements",
        "apply for this job", "apply now"
    ]

    evidence_count = sum(term in text for term in role_terms)

    # Exclude obvious search/category pages
    broad_terms = [
        "internships in india", "jobs in india",
        "job vacancies", "internship snapshot",
        "search results"
    ]

    if any(term in title for term in broad_terms):
        return False

    return evidence_count >= 3

individual_pages = [
    page for page in live_fetch.results
    if is_job_listing(page)
]

print("Fetched pages:", len(live_fetch.results))
print("Potential individual listings:", len(individual_pages))

for page in individual_pages:
    print("\n", page.title)
    print(page.url)
```

```python

from IPython.display import display, HTML
from urllib.parse import urlparse
from html import escape
import re

# Use the live results already collected
pages = {
    p.url.rstrip("/"): p for p in live_fetch.results
}

cards = []

for job in live_results:
    title = (job.get("title") or "Untitled opportunity").strip()
    url = (job.get("url") or "").strip()
    snippet = (job.get("snippet") or "").strip()

    if not url.startswith(("https://", "http://")):
        continue

    page = pages.get(url.rstrip("/"))
    description = snippet or (
        re.sub(r"\s+", " ", page.text)[:300]
        if page and page.text else
        "Open the source page to explore this opportunity."
    )

    domain = urlparse(url).netloc.removeprefix("www.")
    text_to_match = (title + " " + description).lower()

    keywords = [
        "artificial intelligence", "machine learning",
        "data science", "python", "intern"
    ]
    matches = sum(k in text_to_match for k in keywords)
    score = round(matches / len(keywords) * 100)

    cards.append({
        "title": title,
        "url": url,
        "domain": domain,
        "description": description,
        "score": score,
        "fetched": page is not None
    })

# Remove duplicate URLs and sort by keyword relevance
unique = {}
for card in cards:
    unique.setdefault(card["url"].rstrip("/"), card)

cards = sorted(
    unique.values(),
    key=lambda x: x["score"],
    reverse=True
)[:10]

card_html = ""

for job in cards:
    title = escape(job["title"])
    domain = escape(job["domain"])
    description = escape(job["description"][:280])
    url = escape(job["url"], quote=True)
    score = job["score"]
    status = "Page fetched" if job["fetched"] else "Search result"
    status_color = "#047857" if job["fetched"] else "#64748b"

    card_html += f"""
    <div style="background:#fff;border:1px solid #e2e8f0;
        border-radius:16px;padding:20px;margin:12px 0;
        box-shadow:0 3px 12px rgba(15,23,42,.05);
        font-family:Arial,sans-serif;color:#172033;">

      <div style="display:flex;justify-content:space-between;
          gap:12px;align-items:flex-start;flex-wrap:wrap;">

        <div style="flex:1;min-width:200px;">
          <div style="font-size:12px;color:#64748b;
              margin-bottom:8px;">{domain}</div>
          <div style="font-size:18px;font-weight:700;
              line-height:1.4;">{title}</div>
        </div>

        <div style="background:#ecfdf5;color:#047857;
            padding:7px 10px;border-radius:10px;
            font-size:12px;font-weight:700;">
          {score}% keyword relevance
        </div>
      </div>

      <p style="font-size:14px;color:#475569;
          line-height:1.7;margin:14px 0;">
          {description}
        </p>

      <div style="display:flex;justify-content:space-between;
          align-items:center;gap:12px;flex-wrap:wrap;">

        <span style="font-size:12px;color:{status_color};">
          ● {status}
        </span>

        <a href="{url}" target="_blank"
           style="display:inline-block;background:#4f46e5;
           color:white;text-decoration:none;padding:10px 16px;
           border-radius:9px;font-weight:700;font-size:13px;">
          View opportunity ↗
        </a>
      </div>
    </div>
    """

if not cards:
    card_html = """
    <p>No results available. Run the TinyFish Search cell first.</p>
    """

display(HTML(f"""
<div style="font-family:Arial,sans-serif;background:#f8fafc;
    padding:24px;border-radius:20px;color:#172033;">

  <div style="background:linear-gradient(120deg,#312e81,#6366f1);
      color:white;padding:28px;border-radius:18px;margin-bottom:20px;">

    <div style="font-size:12px;letter-spacing:2px;
        font-weight:700;opacity:.85;">TINYFISH × TECHNОJAM</div>

    <h1 style="font-size:30px;margin:12px 0 8px;">
      🚀 ApplyPilot AI
    </h1>

    <p style="font-size:15px;line-height:1.6;margin:0;">
      Discover opportunities. Explore careers. Find your next move.
    </p>
  </div>

  <div style="display:flex;gap:12px;flex-wrap:wrap;margin-bottom:20px;">
    <div style="flex:1;min-width:110px;background:white;
        padding:16px;border-radius:12px;border:1px solid #e2e8f0;">
      <div style="font-size:25px;font-weight:800;">{len(cards)}</div>
      <div style="font-size:12px;color:#64748b;">Unique results</div>
    </div>

    <div style="flex:1;min-width:110px;background:white;
        padding:16px;border-radius:12px;border:1px solid #e2e8f0;">
      <div style="font-size:25px;font-weight:800;">
        {sum(c["fetched"] for c in cards)}
      </div>
      <div style="font-size:12px;color:#64748b;">Pages fetched</div>
    </div>

    <div style="flex:1;min-width:110px;background:white;
        padding:16px;border-radius:12px;border:1px solid #e2e8f0;">
      <div style="font-size:25px;font-weight:800;">Live</div>
      <div style="font-size:12px;color:#64748b;">Search source</div>
    </div>
  </div>

  <h2 style="font-size:21px;margin:22px 0 6px;">
    Opportunities for you
  </h2>
  <p style="font-size:13px;color:#64748b;margin-bottom:16px;">
    Ranked using keyword relevance from your existing live search results.
  </p>

  {card_html}

  <p style="font-size:12px;line-height:1.6;color:#64748b;
      margin-top:20px;">
    Note: Relevance scores are simple keyword scores, not AI confidence.
    Results may be job-board pages rather than individual vacancies.
    Verify availability, eligibility, location and application details
    on the source website before applying.
  </p>
</div>
"""))
```
